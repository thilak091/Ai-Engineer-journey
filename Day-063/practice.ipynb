```python
import torch 
from torch import nn
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader
```

```python
#Challenge 1
train_data = torchvision.datasets.MNIST(root='./data',
                                        train=True,transform=transforms.ToTensor(), 
                                        download=True)
test_data = torchvision.datasets.MNIST(root='./data',
                                        train=False,transform=transforms.ToTensor())
train_loader = DataLoader(dataset=train_data, batch_size=64, shuffle=True)
test_loader = DataLoader(dataset=test_data, batch_size=64, shuffle=False)

iterable = iter(train_loader)
image,label=next(iterable)
print(image.shape)
print(label.shape)
```

```python
#Challenge 2
class NeuralNet(nn.Module):
    def __init__(self,input_size,hidden_size,output_size):
        super(NeuralNet,self).__init__()
        self.l1 = nn.Linear(input_size,hidden_size)
        self.relu = nn.ReLU()
        self.l2 = nn.Linear(hidden_size,output_size)

    def forward(self,X):
        out = self.l1(X)
        out = self.relu(out)
        out = self.l2(out)
        return out

model = NeuralNet(input_size=28*28,hidden_size=128,output_size=10)
# flattening it to 2 dimensions of (batch size, 28*28)
image = image.view(image.shape[0],28*28)
pred = model(image)
print(pred.shape)
```

```python
#Challenge 3
epochs = 5
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(),lr=0.001)
for epoch in range(epochs):
    for i, (image,labels) in enumerate(train_loader):
        image = image.view(image.shape[0],28*28)

        # forward
        y_pred = model(image)

        loss = criterion(y_pred,labels)
        
        #Updation
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        if i%300==0:
            print(f'epoch:{epoch}/{epochs} batch:{i} loss={loss.item()}')
```

```python
#Challenge 4:
n_correct=n_samples=0
with torch.no_grad():
    for i, (image,label) in enumerate(test_loader):
        image=image.view(image.shape[0],28*28)
        y_pred = model(image)
        n_samples +=image.shape[0]
        _, preds = torch.max(y_pred,1)
        n_correct += (preds==label).sum().item()
print(f'Accuracy: {(n_correct/n_samples)*100}')
```
