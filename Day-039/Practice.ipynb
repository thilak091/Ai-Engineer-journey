```python
#Challenge 1
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split , cross_val_score
from sklearn import metrics
import numpy as np

texts = [
    "I love this movie",
    "This movie is amazing",
    "The acting was excellent",
    "I really enjoyed the story",
    "This was a fantastic experience",
    "I hate this movie",
    "This movie is terrible",
    "The acting was horrible",
    "I really disliked the story",
    "This was a boring experience"
]

labels = [
    "positive",
    "positive",
    "positive",
    "positive",
    "positive",
    "negative",
    "negative",
    "negative",
    "negative",
    "negative"
]

tfid = TfidfVectorizer()

X = tfid.fit_transform(texts).toarray()
y = [1 if label == "positive" else 0 for label in labels]

X_train , X_test , y_train , y_test = train_test_split(X , y , test_size = 0.4, random_state = 1)

logreg = LogisticRegression()
logreg.fit(X_train , y_train)
test_pred = logreg.predict(X_test)
print(metrics.accuracy_score(y_test , test_pred))

scores = cross_val_score(logreg , X , y , cv = 5 , scoring = 'accuracy')
print(scores.mean())
```

```python
#Challenge 2
from sklearn.feature_extraction.text import CountVectorizer
tfid = TfidfVectorizer()
X = tfid.fit_transform(texts).toarray()
y = [1 if label == "positive" else 0 for label in labels]

X_train , X_test , y_train , y_test = train_test_split(X , y , test_size = 0.4, random_state = 1)

logreg = LogisticRegression()
logreg.fit(X_train , y_train)
test_pred = logreg.predict(X_test)
print(X)
print("Tfidf Vectorizer Accuracy: ",metrics.accuracy_score(y_test , test_pred))

count = CountVectorizer()
X = count.fit_transform(texts).toarray()
y = [1 if label == "positive" else 0 for label in labels]

X_train , X_test , y_train , y_test = train_test_split(X , y , test_size = 0.4, random_state = 1)

logreg = LogisticRegression()
logreg.fit(X_train , y_train)
test_pred = logreg.predict(X_test)
print(X)
print("Count Vectorizer Accuracy: ",metrics.accuracy_score(y_test , test_pred))
```

```python
#Challenge 3
from sklearn.naive_bayes import MultinomialNB

tfid = TfidfVectorizer()
X = tfid.fit_transform(texts).toarray()
y = [1 if label == "positive" else 0 for label in labels]

X_train , X_test , y_train , y_test = train_test_split(X , y , test_size = 0.4, random_state = 1)

logreg = LogisticRegression()
logreg.fit(X_train , y_train)
test_pred = logreg.predict(X_test)
print("Logistic Regression Accuracy: ",metrics.accuracy_score(y_test , test_pred))

mnb = MultinomialNB()
mnb.fit(X_train , y_train)
test_pred = mnb.predict(X_test)
print("Multinomial NB Accuracy: ",metrics.accuracy_score(y_test , test_pred))
```
