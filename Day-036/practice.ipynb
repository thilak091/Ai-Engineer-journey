```python
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split , cross_val_score
import pandas as pd
from sklearn import metrics
```

```python
svc = SVC(kernel = 'linear' , C = 1)
df = pd.read_csv('2d.csv')

X = df[['feature_1','feature_2']]
y = df['target']

X_train , X_test , y_train , y_test = train_test_split(X , y , test_size = 0.3 , random_state =0)

svc.fit(X_train , y_train)

train_pred = svc.predict(X_train)
test_pred = svc.predict(X_test)

print('Linear')
print(metrics.accuracy_score(y_train , train_pred))
print(metrics.accuracy_score(y_test , test_pred))

scores = cross_val_score(svc , X , y , cv=5 , scoring = 'accuracy')
print(scores.mean())

print(metrics.classification_report(y_test , test_pred))

print(svc.support_ , svc.n_support_)
```

```python
print('RBF')
svc1 = SVC(kernel = 'rbf')
svc1.fit(X_train , y_train)

train_pred1 = svc1.predict(X_train)
test_pred1 = svc1.predict(X_test)

print(metrics.accuracy_score(y_train , train_pred1))
print(metrics.accuracy_score(y_test , test_pred1))

scores = cross_val_score(svc1 , X , y , cv=5 , scoring = 'accuracy')
print(scores.mean())

print(metrics.classification_report(y_test , test_pred1))
```

```python
# C Range
c_val = [0.01 , 0.1 , 1 , 10 , 100]
for c in c_val:
    svc = SVC(kernel = 'linear' , C = c)
    svc.fit(X_train , y_train)

    train_pred = svc.predict(X_train)
    test_pred = svc.predict(X_test)

    print('C=',c)
    print(metrics.accuracy_score(y_train , train_pred))
    print(metrics.accuracy_score(y_test , test_pred))

    scores = cross_val_score(svc , X , y , cv=5 , scoring = 'accuracy')
    print(scores.mean())

    print(metrics.classification_report(y_test , test_pred))

    print(svc.support_ , svc.n_support_)
    print('\n\n')
```

```python
# gamma Range
g_val = [0.001 , 0.01 , 0.1 , 1 , 10]
for g in g_val:
    svc2 = SVC(kernel = 'linear' , gamma = g)
    svc2.fit(X_train , y_train)

    train_pred2 = svc2.predict(X_train)
    test_pred2 = svc2.predict(X_test)

    print('gamma=',g)
    print(metrics.accuracy_score(y_train , train_pred2))
    print(metrics.accuracy_score(y_test , test_pred2))

    scores = cross_val_score(svc2 , X , y , cv=5 , scoring = 'accuracy')
    print(scores.mean())

    print(metrics.classification_report(y_test , test_pred2))

    print(svc2.support_ , svc2.n_support_)
    print('\n\n')
```
