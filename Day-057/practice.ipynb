```python
import torch
```

```python
#Challenge 1
X = torch.tensor(4, dtype=torch.float32,requires_grad = True)

Y= X*3

Z = Y+2

print(f'Output: {Z}')
#Local Gradients
dz = 1+0
dy = 3*1
#Backward propagation and final gradient
dz_dx = dz*dy
print('Manually Calculated:', dz_dx)

#Using AutoGrad 
Z.backward()
print(f'AutoGrad Calc: {X.grad}')
```

```python
#Challenge 2
X = torch.tensor(2,dtype=torch.float32, requires_grad=True)
Z = (3*X)+2
Y = Z**2

#Calculating Local Gradients
dy = 2*Z
dz = (3*1)+0

#Manual Back Propogation and Gradients
dy_dx = dy*dz
print('Output:', Y)
print(f'Manual Calculation:{dy_dx}')

#With AutoGrad
Y.backward()
print(f'Auto Calculation:{X.grad}')
```

```python
#Challenge 3
print("Operation Without Clearing Gradience:")
X = torch.tensor(3,dtype=torch.float32, requires_grad=True)

Y = X*2
Y.backward()
print(f'Gradient After 1st Computation: {X.grad}')

Z= X+3
Z.backward()
print(f'Gradient After 2st Computation: {X.grad}')

print("Operation With Proper Gradience Clearing:")
X = torch.tensor(3,dtype=torch.float32, requires_grad=True)

Y = X*2
Y.backward()
print(f'Gradient After 1st Computation: {X.grad}')
X.grad.zero_()

Z= X+3
Z.backward()
print(f'Gradient After 2st Computation: {X.grad}')
X.grad.zero_()
```

```python
#Challenge 4
W = torch.tensor(0.0, dtype=torch.float32, requires_grad=True)
learning = 0.1

for i in range(20):
    loss = (W-5)**2
    loss.backward()
    with torch.no_grad():
        #Gradient Calc
        dloss_dz = 2*(W-5)*(1)
        W.sub_(learning * dloss_dz)
        W.grad.zero_()
    print(f'Iteration {i+1}:Loss = {loss}: Updated Weights: {W}')
    
```

```python
#Challenge 5
W = torch.tensor(0.0, dtype=torch.float32, requires_grad=True)
learning = 0.1

for i in range(20):
    loss = (W-5)**2
    loss.backward()
    with torch.no_grad():
        W.sub_(learning * W.grad)
        W.grad.zero_()
    print(f'Iteration {i+1}:Loss = {loss}: Updated Weights: {W}')
    
```
