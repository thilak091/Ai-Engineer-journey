```python
#Challenge 1
import pandas as pd
from sklearn.model_selection import train_test_split , cross_val_score
from sklearn.naive_bayes import GaussianNB
from sklearn import metrics

df = pd.read_csv("2d.csv")
X = df[['feature_1' , 'feature_2']]
y = df['target']

X_train , X_test , y_train , y_test = train_test_split(X , y , test_size = 0.25 , random_state = 54)
gnb = GaussianNB()

gnb.fit(X_train , y_train)

train_pred = gnb.predict(X_train)
test_pred = gnb.predict(X_test)

print('Training Accuracy: ',metrics.accuracy_score(y_train , train_pred))
print('Testing Accuracy: ',metrics.accuracy_score(y_test , test_pred))
scores = cross_val_score(gnb , X , y , cv = 5 , scoring = 'accuracy')
print('Cross Val Score: ' , scores.mean())

print("Classification Report:")
print(metrics.classification_report(y_test , test_pred))

```

```python
#Challenge 4
from sklearn.datasets import load_iris

iris = load_iris()

X_new = iris.data
y_new = iris.target
X_train , X_test , y_train , y_test = train_test_split(X_new , y_new , test_size = 0.25 , random_state = 54)
gnb = GaussianNB()

gnb.fit(X_train , y_train)

train_pred = gnb.predict(X_train)
test_pred = gnb.predict(X_test)

print('Training Accuracy: ',metrics.accuracy_score(y_train , train_pred))
print('Testing Accuracy: ',metrics.accuracy_score(y_test , test_pred))
scores = cross_val_score(gnb , X_new , y_new , cv = 5 , scoring = 'accuracy')
print('Cross Val Score: ' , scores.mean())
```
