```python
#Challenge 3
X = torch.tensor([[1], [2], [3], [4], [5]],dtype= torch.float32)
y = torch.tensor([[2], [4], [6], [8], [10]],dtype= torch.float32)

model1 = nn.Linear(n_features, n_features)
loss1 = nn.MSELoss()
optimizer = torch.optim.SGD(model1.parameters(), lr=0.01)
w,b = model1.parameters()
print("Model Parameter before Update:",w[0])
#training loop
for i in range(1):
    #forward pass
    y_pred = model1(X)
    l = loss1(y, y_pred)

    #Backward pass
    l.backward()

    #Update
    optimizer.step()

    optimizer.zero_grad()
w,b = model1.parameters()
print("Model Parameter After Update:",w[0])
```

```python
#Challenge 4 and 5
X = torch.tensor([[-2],[-1],[0],[1],[2]],dtype=torch.float32)
y = torch.tensor([[-3],[-1],[1],[3],[5]],dtype=torch.float32)

model1 = nn.Linear(n_features, n_features)
loss1 = nn.MSELoss()
optimizer = torch.optim.SGD(model1.parameters(), lr=0.01)
w,b = model1.parameters()
print("Model weights before Update:",w[0])
print("Model Bias before Update:",b[0])
#training loop
for i in range(100):
    #forward pass
    y_pred = model1(X)
    l = loss1(y, y_pred)
    if i==0:
        print('Initial Loss:',l.item())

    #Backward pass
    l.backward()

    #Update
    optimizer.step()

    optimizer.zero_grad()
w,b = model1.parameters()
print("Model weights before Update:",w[0])
print("Model Bias before Update:",b[0])

print('Giving the input Values for Final Prediction as:[-2],[-1],[0],[1],[2]')
y_pred = model1(X)
l = loss1(y, y_pred)
print('The Predictions:', model1(X))
print('Final Loss:',l.item())
```
